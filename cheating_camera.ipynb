```python
import cv2
import numpy as np
import traceback
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode, b64encode

# CHEATING LOGIC (YOLO + MediaPipe):
# cheating logic code:
def apply_hybrid_cheating_logic(img, consecutive_cheating_frames):
    phone_detected_alarm = False
    is_looking_away = False

    img_h, img_w, _ = img.shape

    # DETECT PHONE (YOLO):
    phone_results = phone_model(img, verbose=False)
    for r in phone_results:
        for box in r.boxes:
            conf = float(box.conf[0])
            if conf >= 0.40:
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                phone_detected_alarm = True
                cv2.rectangle(img, (x1, y1), (x2, y2), (0, 0, 255), 2)
                cv2.putText(img, f"Phone! {conf:.2f}", (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

    # FIND WHERE PEOPLE ARE SITTING (YOLO):
    person_centers = []
    person_results = person_counter(img, classes=[0], verbose=False)

    for r in person_results:
        for box in r.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            # Find the center X coordinate of this person's body:
            center_x = (x1 + x2) // 2
            person_centers.append(center_x)

    total_people_in_frame = len(person_centers)

    # PROCESS HEAD (MediaPipe):
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    face_results = face_mesh.process(img_rgb)

    if face_results.multi_face_landmarks:
        for face_landmarks in face_results.multi_face_landmarks:
            face_2d = []
            face_3d =[]

            for idx, lm in enumerate(face_landmarks.landmark):
                if idx in[33, 263, 1, 61, 291, 199]:
                    if idx == 1: # Nose tip
                        nose_2d = (int(lm.x * img_w), int(lm.y * img_h))
                    x, y = int(lm.x * img_w), int(lm.y * img_h)
                    face_2d.append([x, y])
                    face_3d.append([x, y, lm.z])

            face_2d = np.array(face_2d, dtype=np.float64)
            face_3d = np.array(face_3d, dtype=np.float64)

            focal_length = 1 * img_w
            cam_matrix = np.array([[focal_length, 0, img_h / 2],[0, focal_length, img_w / 2], [0, 0, 1]])
            dist_matrix = np.zeros((4, 1), dtype=np.float64)

            success, rot_vec, trans_vec = cv2.solvePnP(face_3d, face_2d, cam_matrix, dist_matrix)
            rmat, jac = cv2.Rodrigues(rot_vec)
            angles, mtxR, mtxQ, Qx, Qy, Qz = cv2.RQDecomp3x3(rmat)

            # Get Angles:
            y_angle = angles[1] * 360  # Yaw (Left/Right)
            x_angle = angles[0] * 360  # Pitch (Up/Down)

            nose_x = nose_2d[0]

            # CHECK SURROUNDINGS: Are there bodies to the left or right of THIS face?
            # We use 100 pixels as a buffer so it doesn't accidentally count the person's own body
            people_to_the_left = sum(1 for px in person_centers if px < nose_x - 100)
            people_to_the_right = sum(1 for px in person_centers if px > nose_x + 100)

            # Determine direction of head turn:
            looking_screen_left = y_angle < -15
            looking_screen_right = y_angle > 15
            looking_down = x_angle < -15

            is_cheating_turn = False
            status_text = "Forward"
            dot_color = (0, 255, 0) # Green

            # SPATIAL AWARENESS LOGIC:
            if looking_screen_left:
                if people_to_the_left > 0:
                    is_cheating_turn = True
                    status_text = "Looking at friend (Left)!"
                    dot_color = (0, 165, 255) # Orange
                else:
                    status_text = "Looking away (Safe)"
                    dot_color = (255, 255, 0) # Cyan

            elif looking_screen_right:
                if people_to_the_right > 0:
                    is_cheating_turn = True
                    status_text = "Looking at friend (Right)!"
                    dot_color = (0, 165, 255)
                else:
                    status_text = "Looking away (Safe)"
                    dot_color = (255, 255, 0)

            elif looking_down:
                status_text = "Looking down (Safe)"
                dot_color = (255, 255, 0)

            # Apply text based on status:
            if is_cheating_turn:
                is_looking_away = True

            cv2.putText(img, status_text, (nose_x-50, nose_2d[1]-20), cv2.FONT_HERSHEY_SIMPLEX, 0.6, dot_color, 2)
            cv2.circle(img, nose_2d, 5, dot_color, -1)

    # TIMER:
    if is_looking_away:
        consecutive_cheating_frames += 1
    else:
        consecutive_cheating_frames = max(0, consecutive_cheating_frames - 2)

    # DEBUGGING & ALARMS
    cv2.putText(img, f"People in Frame: {total_people_in_frame}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)
    cv2.putText(img, f"Cheating Timer: {consecutive_cheating_frames}/8", (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2)

    cheating_alarm_active = False
    if phone_detected_alarm or consecutive_cheating_frames >= 8:
        cheating_alarm_active = True

    if cheating_alarm_active:
        cv2.putText(img, "WARNING: CHEATING DETECTED!", (20, 100), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)
        cv2.rectangle(img, (0,0), (img.shape[1], img.shape[0]), (0, 0, 255), 10)

    return img, consecutive_cheating_frames

consecutive_cheating_frames = 0
```

```python
import cv2
import numpy as np
import base64
import mysql.connector
import torch
from flask import Flask, request, jsonify
from flask_socketio import SocketIO, emit
from flask_cors import CORS
from pyngrok import ngrok
from ultralytics import YOLO
import mediapipe as mp
from werkzeug.security import generate_password_hash, check_password_hash

if not hasattr(torch, '_safely_patched'):
    _original_load = torch.load
    torch.load = lambda *args, **kwargs: _original_load(*args, **{**kwargs, 'weights_only': False})
    torch._safely_patched = True

app = Flask(__name__)
CORS(app)

socketio = SocketIO(app, cors_allowed_origins="*", async_mode='threading')

# AIVEN MYSQL CREDENTIALS (SECRET INFO)
DB_CONFIG = {
    #"host": "",
    #"port": ,
    #"user": "",
    #"password": "",
    #"database": "proctor_ai"
}

def get_db_connection():
    return mysql.connector.connect(**DB_CONFIG)

# AI MODELS
phone_model = YOLO("/content/drive/MyDrive/senior_project/phone_model_2/weights/last.pt")
person_counter = YOLO('yolov8n.pt')
mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(min_detection_confidence=0.5, min_tracking_confidence=0.5, max_num_faces=5)

student_tracker = {}
INTERNAL_ID_COUNTER = 1
is_processing = False

@app.route('/api/register', methods=['POST'])
def register():
    data = request.json
    try:
        hashed_pw = generate_password_hash(data['pass'], method='pbkdf2:sha256')
        conn = get_db_connection()
        cursor = conn.cursor()
        cursor.execute(
            "INSERT INTO users (title, first_name, last_name, email, password) VALUES (%s, %s, %s, %s, %s)",
            (data['title'], data['firstName'], data['lastName'], data['email'], hashed_pw)
        )
        conn.commit()
        cursor.close()
        conn.close()
        return jsonify({"success": True, "message": "User registered successfully!"})
    except mysql.connector.IntegrityError:
        return jsonify({"success": False, "message": "Email already exists!"}), 400
    except Exception as e:
        return jsonify({"success": False, "message": str(e)}), 500

@app.route('/api/login', methods=['POST'])
def login():
    data = request.json
    conn = get_db_connection()
    cursor = conn.cursor(dictionary=True)
    cursor.execute("SELECT title, first_name, last_name, password FROM users WHERE email = %s", (data['email'],))
    user = cursor.fetchone()
    cursor.close()
    conn.close()

    if user and check_password_hash(user['password'], data['pass']):
        full_name = f"{user['title']} {user['first_name']} {user['last_name']}"
        return jsonify({"success": True, "name": full_name})
    return jsonify({"success": False, "message": "Invalid email or password."}), 401

@app.route('/api/save_alert', methods=['POST'])
def save_alert():
    data = request.json
    conn = get_db_connection()
    cursor = conn.cursor()
    try:
        cursor.execute('''
            INSERT INTO alerts (id, exam_name, student_name, type, severity, confidence, time, date, detail, snapshot, resolution, proctor_name)
            VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
        ''', (
            data['id'], data['exam_name'], data['student_name'], data['type'],
            data['severity'], data['confidence'], data['time'], data['date'],
            data['detail'], data['snapshot'], data['resolution'], data['proctor_name']
        ))
        conn.commit()
        return jsonify({"success": True})
    except Exception as e:
        print("SQL Save Error:", e)
        return jsonify({"success": False, "error": str(e)})
    finally:
        cursor.close()
        conn.close()

@app.route('/api/get_alerts', methods=['GET'])
def get_alerts():
    proctor = request.args.get('proctor')
    conn = get_db_connection()
    cursor = conn.cursor(dictionary=True)
    try:
        cursor.execute("SELECT * FROM alerts WHERE proctor_name = %s ORDER BY id DESC", (proctor,))
        alerts = cursor.fetchall()
        return jsonify({"success": True, "alerts": alerts})
    except Exception as e:
        return jsonify({"success": False, "error": str(e)})
    finally:
        cursor.close()
        conn.close()

@app.route('/api/update_alert', methods=['POST'])
def update_alert():
    data = request.json
    conn = get_db_connection()
    cursor = conn.cursor()
    try:
        cursor.execute("UPDATE alerts SET resolution = %s WHERE id = %s", (data['resolution'], data['id']))
        conn.commit()
        return jsonify({"success": True})
    except Exception as e:
        return jsonify({"success": False, "error": str(e)})
    finally:
        cursor.close()
        conn.close()

@app.route('/api/delete_alert', methods=['POST'])
def delete_alert():
    data = request.json
    conn = get_db_connection()
    cursor = conn.cursor()
    try:
        cursor.execute("DELETE FROM alerts WHERE id = %s", (data['id'],))
        conn.commit()
        return jsonify({"success": True})
    except Exception as e:
        return jsonify({"success": False, "error": str(e)})
    finally:
        cursor.close()
        conn.close()

def process_frame(img):
    global student_tracker, INTERNAL_ID_COUNTER
    img_h, img_w, _ = img.shape
    person_results = person_counter(img, classes=[0], conf=0.40, verbose=False)
    current_frame_detections = []

    for r in person_results:
        for box in r.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            if (x2 - x1) < 80 or (y2 - y1) < 80: continue
            current_frame_detections.append({'bbox': [x1, y1, x2, y2], 'cx': (x1+x2)//2, 'cy': (y1+y2)//2})

    for tid in list(student_tracker.keys()):
        student_tracker[tid]['missed_frames'] += 1
        if student_tracker[tid]['missed_frames'] > 30: del student_tracker[tid]

    matched_tids = []
    for det in current_frame_detections:
        matched_id = None
        min_dist = 150
        for tid, tdata in student_tracker.items():
            if tid in matched_tids: continue
            if np.sqrt((det['cx'] - tdata['last_cx'])**2 + (det['cy'] - tdata['last_cy'])**2) < min_dist:
                min_dist = np.sqrt((det['cx'] - tdata['last_cx'])**2 + (det['cy'] - tdata['last_cy'])**2)
                matched_id = tid

        if matched_id is None:
            matched_id = INTERNAL_ID_COUNTER
            INTERNAL_ID_COUNTER += 1
            student_tracker[matched_id] = {'timer': 0, 'alert_locked': False}

        student_tracker[matched_id].update({'last_cx': det['cx'], 'last_cy': det['cy'], 'bbox': det['bbox'], 'missed_frames': 0})
        matched_tids.append(matched_id)

    active_students = [tid for tid, tdata in student_tracker.items() if tdata['missed_frames'] == 0]
    active_students.sort(key=lambda tid: student_tracker[tid]['last_cx'])
    all_active_centers = [student_tracker[tid]['last_cx'] for tid in active_students]
    processed_students = []

    for display_index, tid in enumerate(active_students):
        tdata = student_tracker[tid]
        display_name = f"Student {display_index + 1}"
        x1, y1, x2, y2 = tdata['bbox']

        pad_x, pad_y, pad_top = int((x2 - x1) * 0.20), int((y2 - y1) * 0.15), int((y2 - y1) * 0.30)
        crop_x1, crop_y1 = max(0, x1 - pad_x), max(0, y1 - pad_top)
        crop_x2, crop_y2 = min(img_w, x2 + pad_x), min(img_h, y2 + pad_y)
        crop = img[crop_y1:crop_y2, crop_x1:crop_x2]

        if crop.size == 0: continue
        crop_h, crop_w, _ = crop.shape

        phone_results = phone_model(crop, verbose=False)
        max_phone_conf = max([float(b.conf[0]) for r in phone_results for b in r.boxes], default=0.0)
        has_phone = max_phone_conf >= 0.80

        is_looking = False
        face_res = face_mesh.process(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))

        if face_res.multi_face_landmarks:
            face_2d, face_3d = [], []
            for idx, lm in enumerate(face_res.multi_face_landmarks[0].landmark):
                if idx in [33, 263, 1, 61, 291, 199]:
                    face_2d.append([int(lm.x * crop_w), int(lm.y * crop_h)])
                    face_3d.append([int(lm.x * crop_w), int(lm.y * crop_h), lm.z])

            if len(face_2d) == 6:
                success, rot_vec, trans_vec = cv2.solvePnP(np.array(face_3d, dtype=np.float64), np.array(face_2d, dtype=np.float64), np.array([[1*crop_w, 0, crop_h/2], [0, 1*crop_w, crop_w/2], [0, 0, 1]]), np.zeros((4, 1), dtype=np.float64))
                y_angle = cv2.RQDecomp3x3(cv2.Rodrigues(rot_vec)[0])[0][1] * 360

                left_neighbors = sum(1 for cx in all_active_centers if cx < tdata['last_cx'] - 100)
                right_neighbors = sum(1 for cx in all_active_centers if cx > tdata['last_cx'] + 100)
                if (y_angle < -15 and left_neighbors > 0) or (y_angle > 15 and right_neighbors > 0):
                    is_looking = True

        timer, locked = tdata['timer'], tdata['alert_locked']
        timer = timer + 1 if is_looking else max(0, timer - 1)
        if timer == 0 and not has_phone: locked = False

        should_alert, status, color = False, "Safe", "#10b981"
        conf_score = 0

        if has_phone:
            if not locked: should_alert, locked = True, True
            conf_score, status, color = int(max_phone_conf * 100), "Phone Usage", "#ef4444"
        else:
            if timer >= 8:
                if not locked: should_alert, locked = True, True
                conf_score, status, color = min(99, 70 + (timer * 2)), "Cheating (Looking at Friend)", "#ef4444"
            elif timer > 0: status, color = "Looking at Friend...", "#f59e0b"

        _, buf = cv2.imencode('.jpg', crop)
        student_tracker[tid]['timer'], student_tracker[tid]['alert_locked'] = timer, locked
        processed_students.append({"id": display_name, "label": display_name, "status": status, "color": color, "is_cheating": should_alert, "real_conf": conf_score, "timer": timer, "image": f"data:image/jpeg;base64,{base64.b64encode(buf).decode('utf-8')}"})

    return processed_students

@socketio.on('process_frame')
def handle_frame(data):
    global is_processing
    if is_processing: return
    try:
        is_processing = True
        encoded_data = data['image'].split(',')[1]
        img = cv2.imdecode(np.frombuffer(base64.b64decode(encoded_data), np.uint8), cv2.IMREAD_COLOR)
        if img is not None:
            results = process_frame(img)
            emit('processed_data', results)
    except Exception as e:
        print(f"Error handling frame: {e}")
    finally:
        is_processing = False

if __name__ == '__main__':
    ngrok.kill()
    # SECRET TOCKEN HERE -> ngrok.set_auth_token("")
    print(f"\n\n {ngrok.connect(5000).public_url} \n\n")
    socketio.run(app, host='0.0.0.0', port=5000, allow_unsafe_werkzeug=True)
```
